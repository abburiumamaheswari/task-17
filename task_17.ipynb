```python
import pandas as pd

data = {
    'Name': ['Uma', 'Sirisha', 'Akshitha', 'Harika',
             'Navya', 'Akshaya', 'Avanthika'],
    'Age': [18, 19, 18, 20, 19, 18, 20],
    'Marks': [85, 92, 76, 88, 95, 67, 82],
    'Department': ['CSE', 'DS', 'CSE', 'DS',
                   'CSE', 'DS', 'CSE']
}

df = pd.DataFrame(data)

print(df)
```

```python
print(df[df['Marks'] > 80])
```

```python
print(df[df['Age'] == 18])
```

```python
print(df[(df['Marks'] > 80) &
         (df['Age'] < 20)])
```

```python
print(df.sort_values(
    by=['Department', 'Marks'],
    ascending=[True, False]
))
```

```python
filtered = df[df['Marks'] > 80]

print("Filtered Student Records:")
print(filtered)

print("\nTotal Students:", len(filtered))
print("Average Marks:", filtered['Marks'].mean())
print("Highest Marks:", filtered['Marks'].max())
print("Lowest Marks:", filtered['Marks'].min())
```
